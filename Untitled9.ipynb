```python
from nltk import word_tokenize
from nltk.import ngrams
import nltk

# Download tokenizer data if required
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

# Given sentence
sentence = (
    "AI is simulation of human intelligence by machine. "
    "It includes expert systems NLP speech recognition "
    "and machine vision."
)

# Tokenize the sentence
words = word_tokenize(sentence)

print("Original Sentence:")
print(sentence)

print("\nTokens:")
print(words)

# Generate bi-grams
bigrams = list(ngrams(words, 2))

print("\nBi-grams:")
for bg in bigrams:
    print(bg)

# Generate tri-grams
trigrams = list(ngrams(words, 3))

print("\nTri-grams:")
for tg in trigrams:
    print(tg)
```
